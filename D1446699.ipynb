```python
def collect_codes(raw_codes, saved=None):
    # 若未傳入 saved，從空串列開始；若有，則複製一份以防修改原始資料
    result = list(saved) if saved is not None else []

    for code in raw_codes:
        cleaned = code.strip().upper()
        # 略過空字串且確保不重複
        if cleaned and cleaned not in result:
            result.append(cleaned)
    return result

# 建立本題所需的原始資料
saved = ["BASE"]
raw_a = [
    " welcome" + tail + " ",
    "vip" + tail,
    "",
    "WELCOME" + tail,
    "   "
]
raw_b = [
    " new" + tail + " ",
    " welcome" + tail + " "
]

# 複製一份原始 raw_a 供最後印出比對
raw_a_original = list(raw_a)

# 依序計算活動 A 與 B
activity_a = collect_codes(raw_a, saved=saved)
activity_b = collect_codes(raw_b)

# 印出結果
print(f"學號：{student_id}")
print(f"活動A：{activity_a}")
print(f"活動B：{activity_b}")
print(f"原始saved：{saved}")
print(f"原始raw_a：{raw_a_original}")

# 測試空資料
empty_test = collect_codes([])
print(f"空資料：{empty_test}")
```

```python
def cumulative_spend(daily_spend, n):
    # 遞迴終止條件
    if n == 0:
        return 0
    # 遞迴縮小問題：前 n-1 天的費用加上當天 (第 n 天，索引值為 n-1) 的費用
    return cumulative_spend(daily_spend, n - 1) + daily_spend[n - 1]

# 確保變數存在，若未定義則進行輸入或使用預設值
try:
    if 'student_id' not in globals() or not student_id:
        student_id = "D1446699"
except Exception:
    student_id = "D1446699"

# 重新計算 a, b, c, s 以免遺失狀態
a = int(student_id[-3])
b = int(student_id[-2])
c = int(student_id[-1])
s = a + b + c

# 建立每日費用及指定查詢天數 n
daily_spend = [100 + 10 * a, 150 + 10 * b, 0, 200 + 10 * c, 50 + 10 * s]
n = 3 + c % 3

# 印出報告與指定天數累計
print(f"學號：{student_id}")
print(f"每日費用：{daily_spend}")
print(f"指定n：{n}")
print(f"指定前{n}天累計：{cumulative_spend(daily_spend, n)}元")

# 依序印出前 0 至 5 天的累計費用
for i in range(6):
    print(f"前{i}天：{cumulative_spend(daily_spend, i)}元")

# 測試空資料
empty_test = cumulative_spend([], 0)
print(f"空資料：{empty_test}元")
```

```python
def select_clicks(raw_clicks, min_clicks=10):
    # 1. map() 搭配 lambda，去掉每筆文字的頭尾空白
    stripped = map(lambda x: x.strip(), raw_clicks)
    # 2. filter(None, ...)，刪掉空字串 (在 Python 中空字串在 bool 判斷下為 False)
    filtered_empty = filter(None, stripped)
    # 3. map(int, ...) 轉整數，並用 list() 保存結果
    numbers = list(map(int, filtered_empty))

    # 4. 使用 filter() 搭配 lambda 篩選大於或等於 min_clicks 的項目
    qualified = list(filter(lambda x: x >= min_clicks, numbers))

    return (numbers, qualified)

# 確保變數存在，若未定義則進行輸入或使用預設值
try:
    if 'student_id' not in globals() or not student_id:
        student_id = "D1446699"
except Exception:
    student_id = "D1446699"

# 重新計算 a, b, c
a = int(student_id[-3])
b = int(student_id[-2])
c = int(student_id[-1])

# 建立自己的資料
raw_clicks = [str(10 + a), "0", "", " " + str(5 + b) + " ", "   ", str(15 + c), str(5 + b)]
min_clicks = 5 + b

# 呼叫函式計算
numbers, qualified = select_clicks(raw_clicks, min_clicks=min_clicks)
qualified_count = len(qualified)
qualified_sum = sum(qualified)

# 印出基本結果
print(f"學號：{student_id}")
print(f"點擊門檻：{min_clicks}")
print(f"清理結果：{numbers}")
print(f"合格結果：{qualified}")
print(f"合格筆數：{qualified_count}")
print(f"合格合計：{qualified_sum}")

# 使用 sorted 與 enumerate 排列名次並印出
sorted_qualified = sorted(qualified, reverse=True)
for rank, clicks in enumerate(sorted_qualified, start=1):
    print(f"第{rank}名：{clicks}次")

# 使用門檻 100 呼叫
numbers_100, qualified_100 = select_clicks(raw_clicks, min_clicks=100)
qualified_100_count = len(qualified_100)
qualified_100_sum = sum(qualified_100)

print(f"門檻100合格結果：{qualified_100}")
print(f"門檻100合格筆數：{qualified_100_count}")
print(f"門檻100合格合計：{qualified_100_sum}")
if not qualified_100:
    print("無符合門檻的素材")
```
